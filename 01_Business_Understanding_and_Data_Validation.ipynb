## 2. Missing Value Analysis

**Business objective:** Identify missing data and determine whether it represents a data-quality problem or a legitimate business situation.

We will calculate the number and percentage of missing values in every column before deciding how to handle them.


```python
# Missing value analysis

missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

display(missing_report)
```

## 3. Duplicate Record Analysis

**Business objective:** Identify exact duplicate rows and repeated booking IDs.

Repeated booking IDs require investigation, but they are not automatically errors. We must verify the data before removing anything.


```python
# Check exact duplicate rows

duplicate_rows = df.duplicated().sum()

# Check duplicate Booking IDs

duplicate_booking_ids = df["Booking ID"].duplicated().sum()

print("Exact duplicate rows:", duplicate_rows)
print("Repeated Booking IDs:", duplicate_booking_ids)
print("Total rows:", len(df))
print("Unique Booking IDs:", df["Booking ID"].nunique())
```

## . Duplicate Booking ID Investigation

### Business objective

Investigate why some Booking IDs appear more than once.

### Questions

* Are the repeated records identical in their business information?
* Do repeated Booking IDs have different booking statuses?
* Are the repeated IDs associated with different customers?
* Should these records be retained or investigated further?

We will inspect the records before making any changes.


```python
# Identify all rows with repeated Booking IDs

duplicate_booking_mask = df["Booking ID"].duplicated(keep=False)

duplicate_booking_df = df.loc[
    duplicate_booking_mask
].sort_values("Booking ID")

print("Rows belonging to repeated Booking IDs:",
      len(duplicate_booking_df))

display(duplicate_booking_df.head(20))
```

```python
# Count the number of rows for each repeated Booking ID

repeated_id_counts = (
    duplicate_booking_df
    .groupby("Booking ID")
    .size()
    .sort_values(ascending=False)
)

print("Number of repeated Booking IDs:",
      repeated_id_counts.size)

print("\nMost frequently repeated Booking IDs:")
display(repeated_id_counts.head(10))
```

```python
# Count distinct booking statuses for each repeated Booking ID

status_variation = (
    duplicate_booking_df
    .groupby("Booking ID")["Booking Status"]
    .nunique()
)

print(
    "Repeated IDs with multiple booking statuses:",
    (status_variation > 1).sum()
)

print(
    "Repeated IDs with only one booking status:",
    (status_variation == 1).sum()
)
```

```python
# Count distinct customers associated with each repeated Booking ID

customer_variation = (
    duplicate_booking_df
    .groupby("Booking ID")["Customer ID"]
    .nunique()
)

print(
    "Repeated IDs linked to multiple customers:",
    (customer_variation > 1).sum()
)

print(
    "Repeated IDs linked to one customer:",
    (customer_variation == 1).sum()
)
```

```python
# Investigate Payment Method missingness by Booking Status

payment_missing_summary = (
    df.assign(
        Payment_Method_Missing=df["Payment Method"].isna()
    )
    .groupby("Booking Status")["Payment_Method_Missing"]
    .agg(
        Total_Bookings="size",
        Missing_Payment_Method="sum"
    )
)

payment_missing_summary["Missing_Percentage"] = (
    payment_missing_summary["Missing_Payment_Method"]
    / payment_missing_summary["Total_Bookings"]
    * 100
).round(2)

display(payment_missing_summary)
```

#— Booking Outcome Consistency

## Business Objective

Verify whether the cancellation and incomplete-ride indicators agree with the Booking Status column.

## Business Questions

1. Does every customer-cancelled booking have a customer cancellation indicator of 1?
2. Does every driver-cancelled booking have a driver cancellation indicator of 1?
3. Does every incomplete booking have an incomplete-ride indicator of 1?
4. Are any indicators incorrectly set to 1 for other booking statuses?

These checks will help identify inconsistent records before data cleaning.


```python
# Check booking status against cancellation indicators

checks = {
    "Customer Cancellation": (
        (df["Booking Status"] == "Cancelled by Customer")
        == (df["Cancelled Rides by Customer"] == 1)
    ),

    "Driver Cancellation": (
        (df["Booking Status"] == "Cancelled by Driver")
        == (df["Cancelled Rides by Driver"] == 1)
    ),

    "Incomplete Ride": (
        (df["Booking Status"] == "Incomplete")
        == (df["Incomplete Rides"] == 1)
    )
}

for name, result in checks.items():
    print(f"\n{name}")
    print("Matching records:", result.sum())
    print("Mismatching records:", (~result).sum())
```

Check numerical business rules

```python
# Check potentially invalid numerical values

print("Negative Booking Values:",
      (df["Booking Value"] < 0).sum())

print("Negative Ride Distances:",
      (df["Ride Distance"] < 0).sum())

print("Negative Avg VTAT:",
      (df["Avg VTAT"] < 0).sum())

print("Negative Avg CTAT:",
      (df["Avg CTAT"] < 0).sum())

print("Driver Ratings outside 0-5:",
      (~df["Driver Ratings"].between(0, 5)).sum())

print("Customer Ratings outside 0-5:",
      (~df["Customer Rating"].between(0, 5)).sum())
```

Investigate zero values by Booking Status

```python
# Investigate zero values by booking outcome

zero_summary = (
    df.groupby("Booking Status")
    .agg(
        Total_Bookings=("Booking ID", "size"),

        Zero_Booking_Value=(
            "Booking Value",
            lambda x: (x == 0).sum()
        ),

        Zero_Ride_Distance=(
            "Ride Distance",
            lambda x: (x == 0).sum()
        ),

        Zero_Driver_Rating=(
            "Driver Ratings",
            lambda x: (x == 0).sum()
        ),

        Zero_Customer_Rating=(
            "Customer Rating",
            lambda x: (x == 0).sum()
        )
    )
)

display(zero_summary)
```
